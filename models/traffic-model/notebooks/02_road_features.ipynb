# 02 — Road Feature Extraction
Extract OSM road features + capacity proxy + network topology.

```python
import sys; sys.path.insert(0, '../src')
import osm_loader, osm_features, road_network, config

roads, nodes, stats = osm_loader.load_osm()
roads = osm_features.compute_base_features(roads)
roads = road_network.add_topology_features(roads)

print('Road features extracted:')
print(roads[['road_id','road_type','road_length_m','lane_count','speed_limit_kmh',
             'is_oneway','road_capacity_proxy','intersection_density','node_degree',
             'connected_road_count','road_hierarchy']].head(5).to_string(index=False))

print(f'\nDrivable segments: {roads.is_drivable.sum()} / {len(roads)}')
print(f'Total network: {roads.road_length_m.sum()/1000:.0f} km')

if config.OSM_ROADS_GPKG.exists():
    import geopandas as gpd
    saved = gpd.read_file(config.OSM_ROADS_GPKG)
    print(f'Saved to {config.OSM_ROADS_GPKG.name}: {len(saved)} rows')
```
