# 03 — Synthetic Traffic Generation
Generate synthetic traffic observations conditioned on real OSM road attributes.

**DISCLAIMER:** This is SYNTHETIC data for pipeline prototyping, not real traffic measurements.

```python
import sys; sys.path.insert(0, '../src')
import traffic_generator as tg
import config

traffic = tg.generate()
print(f'\nDataset shape: {traffic.shape}')
print(f'Date range: {traffic.timestamp.min()} to {traffic.timestamp.max()}')
print(f'Road segments: {traffic.road_id.nunique()}')
print(f'\nTraffic volume stats:')
print(traffic['traffic_volume'].describe().round(1).to_string())

print(f'\nMean by road type:')
print(traffic.groupby('road_type')['traffic_volume'].mean().sort_values(ascending=False).round(0).to_string())
```
