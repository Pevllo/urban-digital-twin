# 01 — OSM Data Inspection
Load and inspect the raw OSM road network data.

```python
import sys; sys.path.insert(0, '../src')
import osm_loader, config, pandas as pd
pd.set_option('display.max_colwidth', 80)

roads, nodes, stats = osm_loader.load_osm()
print('=== Loading ===')
for k, v in stats.items(): print(f'  {k}: {v}')

print('\n=== Inspection Report ===')
report = osm_loader.inspect_osm(roads, stats)
print(report.to_string(index=False))

print('\n=== Highway types ===')
print(roads['highway'].value_counts().to_string())

print('\n=== Sample rows ===')
print(roads.head(3).to_string())
```
