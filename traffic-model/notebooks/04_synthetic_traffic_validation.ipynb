# 04 — Synthetic Traffic Validation
Validate the generated synthetic traffic dataset.

```python
import sys; sys.path.insert(0, '../src')
import osm_loader, osm_features, road_network, traffic_validation, config
import pandas as pd
import warnings; warnings.filterwarnings('ignore')

roads, nodes, stats = osm_loader.load_osm()
roads = osm_features.compute_base_features(roads)
roads = road_network.add_topology_features(roads)
adjacency = road_network.build_road_adjacency(roads[roads.is_drivable].reset_index(drop=True))

traffic_gdf = __import__('geopandas').read_file(config.SYNTHETIC_TRAFFIC_GPKG)

params = {'formula': 'See synthetic_data_report.md', 'config': {'RANDOM_SEED': config.RANDOM_SEED}}
validation = traffic_validation.validate_and_report(roads, traffic_gdf, adjacency, params)
print('\nValidation results:')
print(validation.to_string(index=False))
```
